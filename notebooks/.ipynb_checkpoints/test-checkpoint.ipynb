```python
import numpy as np
from ann_engine.core import Tensor, Parameter
from ann_engine.layers import ReLU 
```

```python
def test_relu():
    """Simple ReLU test"""
    print("=" * 50)
    print("TESTING RELU ACTIVATION")
    print("=" * 50)
    
    try:
        # Create ReLU
        relu = ReLU()
        print(f"1. Created ReLU: {relu}")
        
        # Test data
        x = Parameter(np.array([-2.0, -1.0, 0.0, 1.0, 2.0]))
        print(f"\n2. Input: {x.data}")
        
        # Forward pass
        out = relu(x)
        print(f"3. Output: {out.data}")
        print(f"   Expected: [0. 0. 0. 1. 2.]")
        
        # Backward pass
        loss = out.sum()
        loss.backward()
        print(f"\n4. Gradients: {x.grad}")
        print(f"   Expected: [0. 0. 0. 1. 1.]")
        
        # Verify
        assert np.allclose(out.data, [0, 0, 0, 1, 2]), "Forward pass wrong"
        assert np.allclose(x.grad, [0, 0, 0, 1, 1]), "Backward pass wrong"
        
        print("\n✅ ALL TESTS PASSED!")
        
    except Exception as e:
        print(f"\n❌ ERROR: {e}")
        import traceback
        traceback.print_exc()
    
    print("\n" + "=" * 50)
    input("Press Enter to exit...")
```

```python
if __name__ == "__main__":
    test_relu()
```

```python
import numpy as np
from ann_engine.core import Tensor, Parameter
from ann_engine.layers.dense import Dense
from ann_engine.layers import ReLU
from ann_engine.losses.loss import MSELoss
from ann_engine.optimizers import SGD
```

```python
np.random.seed(42)
X_data = np.random.randn(1000,2)
y_data = 2 * X_data[:, 0:1] + 3 * X_data[:, 1:2] + 0.1 * np.random.randn(1000, 1)
```

```python
X = Tensor(X_data)
y_true = Tensor(y_data)
```

```python
model = Dense(2,1, bias=True, initialization='xavier')
print(f"model:{model}")
print(f"initial weights:{model.W.data.flatten()}")
print(f"initial bias:{model.b.data.flatten()}")
```

```python
criterion =  MSELoss()
optimizer = SGD(model.parameters(), lr=0.01)
```

```python
print("\nTrainning...")
for epoch in range(100):
    y_pred = model.forward(X)
    loss = criterion(y_pred, y_true)
    # backward
    optimizer.zero_grad()
    loss.backward()
    # Update
    optimizer.step()

    if epoch % 20 == 0:
        print(f"Epoch {epoch:3d}, loss : {loss.data:.6f}")


print(f"\nFinal weights: {model.W.data.flatten()}")
print(f"Final bias: {model.b.data.flatten()}")
print(f"Expected weights: [2.0, 3.0]")
print(f"Expected bias: [0.0]")
    
```
