```python
import numpy as np
from ann_engine.layers.layers import Dense
from ann_engine.layers import Sequential
from ann_engine.optimizers import Adam
```

```python
X_train = np.random.randn(1000, 784)
y_train = np.random.randint(0, 10, (1000, 10)) 
```

```python
model1 = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
print(model1)
```

```python
model1.compile(optimizer=Adam(lr=0.01),
             loss = 'categorical_crossentropy',
             metrics = ['accuracy'])
```

```python
history_of_ann_engine = model1.fit(X_train, y_train, epochs=100, batch_size=32, verbose=1)
```

```python
model1.summary()
```

```python
np.random.seed(42)
```

```python
def make_spiral(n_samples=1000, n_classes=3, noise=0.2):
    """Create spiral dataset for multi-class classification"""
    X = np.zeros((n_samples * n_classes, 2))
    y = np.zeros((n_samples * n_classes, n_classes))
    
    for class_id in range(n_classes):
        ix = range(n_samples * class_id, n_samples * (class_id + 1))
        r = np.linspace(0.0, 1, n_samples)
        t = np.linspace(class_id * 4, (class_id + 1) * 4, n_samples) + np.random.randn(n_samples) * noise
        X[ix] = np.c_[r * np.sin(t), r * np.cos(t)]
        y[ix, class_id] = 1
    
    return X, y

X, y = make_spiral(n_samples=300, n_classes=3, noise=0.2)
print(f"X shape: {X.shape}, y shape: {y.shape}")
print(f"X range: [{X.min():.2f}, {X.max():.2f}]")
print(f"Sample classes: {np.argmax(y[:10], axis=1)}")
```
