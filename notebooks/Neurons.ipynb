```python
# basic of neurons
inputs = [1, 2, 3,2.5]
weights = [0.2, 0.8, -0.5, 1.0]
bisas = 2
output = []

for i in range(len(inputs)):
    output2 = inputs[i]*weights[i]
    output.append(output2)
sum(output)+bisas
```
